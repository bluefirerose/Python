Business Question Analysis 1

1. In which country do cyberattacks commonly occur?

```python
# Count cyberattacks by country
country_attacks = df['Country'].value_counts()

# Total number of cyberattacks
total_attacks = country_attacks.sum()

# Create horizontal bar graph
plt.figure(figsize=(10, 6))

sns.barplot(
    x=country_attacks.values,
    y=country_attacks.index
)

plt.title('Cyberattacks by Country')
plt.xlabel('Number of Cyberattacks')
plt.ylabel('Country')

plt.tight_layout()
plt.show()


# Reference list outside the graph
print("Cyberattacks by Country:")
print("-" * 35)

for country, count in country_attacks.items():
    print(f"{country:<20} {count}")

print("-" * 35)
print(f"{'Total Cyberattacks:':<20} {total_attacks}")

# Summary
print("\nMost affected country:", country_attacks.idxmax())
print("Number of attacks:", country_attacks.max())
```

The graph shows the 10 countries where cyberattacks were identified in the dataset between 2015 and 2024. The UK had the highest number of cyberattacks during this 10-year period. Followed by Brazil, India, Japan and France. This graph helps determine which countries were targeted the most by cyberattacks over the past ten years.

Business Question Analysis 2

2. Does technological advancement greatly impact the growth of cyberattacks?

```python
#Group the attack by year
attacks_year = df.groupby('Year').size()

#Create the line graph
plt.figure(figsize=(10, 6))

plt.plot(
    attacks_year.index,
    attacks_year.values,
    marker='o',
    linewidth=2
)

plt.title('Trend of Cyberattacks from 2015 to 2024')
plt.xlabel('Year')
plt.ylabel('Number of Cyberattacks')

plt.xticks(attacks_year.index)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

#Compute the correlation between the year and number of attacks
correlation = np.corrcoef(
    attacks_year.index,
    attacks_year.values
)[0, 1]

print("Year vs Cyberattack Frequency Correlation:",
      round(correlation, 3))
```

Based on the general interpretation of correlation, the 0.401 has a moderate positive correlation coefficient. The graph shows threat count flactuate instead of gradually increasing every year. This means that technology advancement does not immediately mean growth of cybersecurity attack based on the dataset.


Business Question Analysis 3

3. Which industry is most commonly targeted by cyberattacks?

```python
# Count cyberattacks per industry
industry_attacks = df['Target Industry'].value_counts()

# Total number of attacks
total_attacks = industry_attacks.sum()

# Function to display percentage and actual count
def show_count_and_percentage(pct):
    count = int(round(pct / 100 * total_attacks))
    return f'{pct:.1f}%\n({count})'


plt.figure(figsize=(10, 8))

plt.pie(
    industry_attacks.values,
    labels=industry_attacks.index,
    autopct=show_count_and_percentage,
    startangle=90,
    pctdistance=0.80
)

# Create donut hole
centre_circle = plt.Circle(
    (0, 0),
    0.55,
    fc='white'
)

plt.gca().add_artist(centre_circle)

plt.title(
    f'Distribution of Cyberattacks by Target Industry\n'
    f'Total Cyberattacks: {total_attacks:,}'
)

plt.axis('equal')
plt.tight_layout()
plt.show()


# Display the exact count per industry
print("Cyberattacks per Industry:")
print(industry_attacks)

print("\nTotal Cyberattacks:", total_attacks)




plt.figure(figsize=(14, 7))

sns.lineplot(
    data=industry_year,
    x='Year',
    y='Number of Attacks',
    hue='Target Industry',
    marker='o'
)

plt.title('Cyberattacks by Target Industry per Year')
plt.xlabel('Year')
plt.ylabel('Number of Cyberattacks')
plt.xticks(sorted(df['Year'].unique()))

plt.legend(
    title='Target Industry',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

The IT industry shows the highest cyberattacks per industry for the past ten years, with 15.9% total cyberattacks. This means IT industries should strengthen their security posture to mitigate the risk.

4. How much financial loss is caused by cyberattacks each year?

```python
financial_loss_year = (
    df.groupby('Year')['Financial Loss (in Million $)']
    .sum()
    .sort_index()
)

total_loss = financial_loss_year.sum()

def show_loss(pct):
    value = pct / 100 * total_loss
    return f'{pct:.1f}%\n${value:,.0f}M'


plt.figure(figsize=(11, 8))

plt.pie(
    financial_loss_year.values,
    labels=financial_loss_year.index,
    autopct=show_loss,
    startangle=90
)

plt.title(
    f'Financial Loss from Cyberattacks by Year\n'
    f'Total Financial Loss: ${total_loss / 1000:,.2f} Billion'
)

plt.axis('equal')
plt.tight_layout()
plt.show()
```

Approximately $151.48 billion loss for between 2015 to 2024. This shows how cyberattack costs each year for the whole industr.

5. What is the most common threat type used by hackers?

```python
attack_types = (
    df['Attack Type']
    .value_counts()
    .sort_values()
)

plt.figure(figsize=(10, 6))

plt.hlines(
    y=attack_types.index,
    xmin=0,
    xmax=attack_types.values,
    linewidth=2
)

plt.scatter(
    attack_types.values,
    attack_types.index,
    s=100
)

plt.title('Frequency of Cyberattack Types')
plt.xlabel('Number of Cyberattacks')
plt.ylabel('Attack Type')

plt.tight_layout()
plt.show()


attack_type_counts = df['Attack Type'].value_counts()

print(attack_type_counts)
```

Here, we determined the known threat types used for security attacks. Distributed Denial of Service (DDoS) scored the highest with 531. This affects the system's availability for use. Phishing, with 529, ranked second, as with the boom of AI, there are tools that empower these attacks to be smarter or more presentable than before.

6. What is the most common attack vector used to penetrate a company’s security?

```python
# Count each attack source per year
attack_source_year = pd.crosstab(
    df['Year'],
    df['Attack Source']
)

print(attack_source_year)

# Create graph
ax = attack_source_year.plot(
    kind='bar',
    figsize=(17, 7),
    width=0.8
)

# Add count above each bar
for container in ax.containers:
    ax.bar_label(
        container,
        fontsize=8,
        padding=2
    )

plt.title('Cyberattack Sources per Year')
plt.xlabel('Year')
plt.ylabel('Number of Cyberattacks')

plt.xticks(rotation=0)

plt.legend(
    title='Attack Source',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

```python
The above graph the total number of attack source per year. 
```

```python
attack_source_counts = df['Attack Source'].value_counts()

print(attack_source_counts)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=attack_source_counts.values,
    y=attack_source_counts.index
)

plt.title('Common Sources of Cyberattacks')
plt.xlabel('Number of Attacks')
plt.ylabel('Attack Source')

plt.show()
```

7. What are the most common types of security vulnerabilities?

```python
# Count vulnerabilities per year
vulnerability_year = (
    df.groupby(['Year', 'Security Vulnerability Type'])
      .size()
      .reset_index(name='Number of Incidents')
)

# Display the data
print(vulnerability_year)

# Create scatter plot
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=vulnerability_year,
    x='Year',
    y='Number of Incidents',
    hue='Security Vulnerability Type',
    s=150
)

plt.title('Security Vulnerabilities by Year')
plt.xlabel('Year')
plt.ylabel('Number of Incidents')

plt.xticks(sorted(df['Year'].unique()))

plt.legend(
    title='Security Vulnerability Type',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

```python
vulnerability_counts = df[
    'Security Vulnerability Type'
].value_counts()

print(vulnerability_counts)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=vulnerability_counts.values,
    y=vulnerability_counts.index
)

plt.title('Common Security Vulnerabilities')
plt.xlabel('Number of Incidents')
plt.ylabel('Security Vulnerability')

plt.show()
```

Zero-day vulnerability has been determined in our dataset as the most common security vulnerability, possibly due to the cost of software and tool updates.

8. What security tools are commonly used by industries?

```python
defense_counts = df[
    'Defense Mechanism Used'
].value_counts()

print(defense_counts)

security_tools = pd.crosstab(
    df['Target Industry'],
    df['Defense Mechanism Used']
)

security_tools


plt.figure(figsize=(12, 7))

sns.heatmap(
    security_tools,
    annot=True,
    fmt='d',
    cmap='YlGnBu'
)

plt.title('Security Tools Used Across Industries')
plt.xlabel('Defense Mechanism')
plt.ylabel('Industry')

plt.tight_layout()
plt.show()
```

Business Question Analysis 9



9. How many users are affected by cyberattacks?


```python
users_year = (
    df.groupby('Year')['Number of Affected Users']
    .sum()
)

total_users = df['Number of Affected Users'].sum()

print("Total affected-user instances:",
      f"{total_users:,}")

plt.figure(figsize=(10, 5))

plt.fill_between(
    users_year.index,
    users_year.values,
    alpha=0.4
)

plt.plot(
    users_year.index,
    users_year.values,
    marker='o'
)

plt.title(
    f'Affected Users from Cyberattacks by Year\n'
    f'Total: {total_users:,}'
)

plt.xlabel('Year')
plt.ylabel('Number of Affected Users')

plt.xticks(users_year.index)

plt.tight_layout()
plt.show()
```

The graph shows how many users are affected due to cyberattacks. This helps organizations to improve user awareness and how not to be the weakest link in the organization.

Business Question Analysis 10

10. How does resolution time affect financial loss?

```python
correlation = df[
    'Incident Resolution Time (in Hours)'
].corr(
    df['Financial Loss (in Million $)']
)

print(
    "Correlation:",
    round(correlation, 3)
)
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x='Incident Resolution Time (in Hours)',
    y='Financial Loss (in Million $)',
    scatter_kws={'alpha': 0.4}
)

plt.title(
    'Incident Resolution Time vs Financial Loss'
)

plt.xlabel('Incident Resolution Time (Hours)')
plt.ylabel('Financial Loss (Million $)')

plt.show()
```

Correlation between resolution time and finacial loss got a result of -0.013. This means that greater time resolution does not mean greater financial loss. There is no correlation basing on the synthetic dataset we have used. 

Conclusion

Overall, this exploratory data analysis showed that the UK was the most affected country by cyberattacks, while the IT industry experienced the highest number of attacks over the past 10 years. A moderate positive correlation of 0.401 was identified between year and cyberattack frequency, indicating a general tendency for cyberattacks to increase over time. However, the dataset does not directly measure technological advancement; therefore, it cannot be concluded that technological advancement affects the growth of cyberattacks.

The analysis also identified approximately $151.48 billion in total financial losses caused by cyberattacks. Distributed Denial-of-Service (DDoS) was the most common type of attack and can disrupt business operations by affecting the availability of systems and services. Nation-state actors were the most common attack source, while zero-day vulnerabilities were the most common security vulnerability. Antivirus software was the most commonly used defense mechanism for protecting computer devices.

Furthermore, a total of 1,514,052,409 affected-user instances were recorded over the 10-year period. Finally, the analysis found almost no correlation between incident resolution time and financial loss, suggesting that resolution time alone does not have a meaningful linear relationship with the financial losses recorded in this dataset.

Limitation

The dataset used in this exploratory data analysis is synthetic and intended for educational and research purposes. Therefore, the results should not be interpreted as an exact representation of real-world global cybersecurity incidents. Using reliable real-world cybersecurity data in future studies could provide more realistic findings and may support organizations in improving their Intrusion Detection Systems (IDS), Intrusion Prevention Systems (IPS), and overall cybersecurity posture.

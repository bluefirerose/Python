Project Brief
The rapid growth of Artificial Intelligence (AI) has affected many industries. With these technological advancements, cybersecurity measures need to continuously evolve and improve. This project examines global cybersecurity threats affecting industries. It aims to provide companies with insights into the most common and widely used types of cyberattacks, as well as the industries most frequently targeted. The dataset used in this project was obtained from Kaggle. The data is synthetic and is intended for educational, research, and data analytics purposes. Soundankar, A. (2025) Global Cybersecurity Threats (2015–2024). Kaggle. Available at: https://www.kaggle.com/datasets/atharvasoundankar/global-cybersecurity-threats-2015-2024/discussion/589979 (Accessed: 8 September 2026)

Business Context
This project focuses on global industries that operate using computer systems and digital technologies, particularly industries that face a high risk of financial loss in the event of a cyberattack. The analysis can help organizations become more aware of cybersecurity threats and identify areas where they can improve their overall security posture.

The project will examine the geographical locations of cyberattacks, the year in which they occurred, threat types, attack vectors, targeted industries, and their financial impact.

Explanatory Data Analysis

In which country do cyberattacks commonly occur?

```python
country_attacks = df['Country'].value_counts()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=country_attacks.values,
    y=country_attacks.index
)

plt.title('Cyberattacks by Country')
plt.xlabel('Number of Cyberattacks')
plt.ylabel('Country')

plt.tight_layout()
plt.show()


print("Most affected country:", country_attacks.idxmax())
print("Number of attacks:", country_attacks.max())
```

Conclusion:






Does technological advancement greatly impact the growth of cyberattacks?

```python
attacks_year = df.groupby('Year').size()

plt.figure(figsize=(10, 6))

plt.plot(
    attacks_year.index,
    attacks_year.values,
    marker='o',
    linewidth=2
)

plt.title('Trend of Cyberattacks from 2015 to 2024')
plt.xlabel('Year')
plt.ylabel('Number of Cyberattacks')

plt.xticks(attacks_year.index)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


correlation = np.corrcoef(
    attacks_year.index,
    attacks_year.values
)[0, 1]

print("Year vs Cyberattack Frequency Correlation:",
      round(correlation, 3))
```

Conclusion





Which industry is most commonly targeted by cyberattacks?

```python
# Count cyberattacks per industry
industry_attacks = df['Target Industry'].value_counts()

# Total number of attacks
total_attacks = industry_attacks.sum()

# Function to display percentage and actual count
def show_count_and_percentage(pct):
    count = int(round(pct / 100 * total_attacks))
    return f'{pct:.1f}%\n({count})'


plt.figure(figsize=(10, 8))

plt.pie(
    industry_attacks.values,
    labels=industry_attacks.index,
    autopct=show_count_and_percentage,
    startangle=90,
    pctdistance=0.80
)

# Create donut hole
centre_circle = plt.Circle(
    (0, 0),
    0.55,
    fc='white'
)

plt.gca().add_artist(centre_circle)

plt.title(
    f'Distribution of Cyberattacks by Target Industry\n'
    f'Total Cyberattacks: {total_attacks:,}'
)

plt.axis('equal')
plt.tight_layout()
plt.show()


# Display the exact count per industry
print("Cyberattacks per Industry:")
print(industry_attacks)

print("\nTotal Cyberattacks:", total_attacks)




plt.figure(figsize=(14, 7))

sns.lineplot(
    data=industry_year,
    x='Year',
    y='Number of Attacks',
    hue='Target Industry',
    marker='o'
)

plt.title('Cyberattacks by Target Industry per Year')
plt.xlabel('Year')
plt.ylabel('Number of Cyberattacks')
plt.xticks(sorted(df['Year'].unique()))

plt.legend(
    title='Target Industry',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

Conclusion





How much financial loss is caused by cyberattacks each year?

```python
# Calculate total financial loss for each year
financial_loss_year = (
    df.groupby('Year')['Financial Loss (in Million $)']
      .sum()
      .sort_index()
)

# Calculate overall financial loss
total_loss = financial_loss_year.sum()

print("Financial Loss Per Year:")
print(financial_loss_year)

print("\nTotal Financial Loss:")
print(f"${total_loss:,.2f} Million")

print("\nTotal Financial Loss in Billions:")
print(f"${total_loss / 1000:,.2f} Billion")

plt.figure(figsize=(10, 8))

plt.pie(
    financial_loss_year.values,
    labels=financial_loss_year.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title(
    f'Financial Loss from Cyberattacks by Year\n'
    f'Total Loss: ${total_loss / 1000:,.2f} Billion'
)

plt.axis('equal')
plt.show()
```

Conlusion

Approximately $151.48 billion loss for between 2015 to 2024

What is the most common threat type used by hackers?

```python
attack_types = (
    df['Attack Type']
    .value_counts()
    .sort_values()
)

plt.figure(figsize=(10, 6))

plt.hlines(
    y=attack_types.index,
    xmin=0,
    xmax=attack_types.values,
    linewidth=2
)

plt.scatter(
    attack_types.values,
    attack_types.index,
    s=100
)

plt.title('Frequency of Cyberattack Types')
plt.xlabel('Number of Cyberattacks')
plt.ylabel('Attack Type')

plt.tight_layout()
plt.show()


attack_type_counts = df['Attack Type'].value_counts()

print(attack_type_counts)
```

Conclusion



What is the most common attack vector used to penetrate a company’s security?

```python
# Count each attack source per year
attack_source_year = pd.crosstab(
    df['Year'],
    df['Attack Source']
)

print(attack_source_year)

# Create graph
ax = attack_source_year.plot(
    kind='bar',
    figsize=(17, 7),
    width=0.8
)

# Add count above each bar
for container in ax.containers:
    ax.bar_label(
        container,
        fontsize=8,
        padding=2
    )

plt.title('Cyberattack Sources per Year')
plt.xlabel('Year')
plt.ylabel('Number of Cyberattacks')

plt.xticks(rotation=0)

plt.legend(
    title='Attack Source',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

```python
attack_source_counts = df['Attack Source'].value_counts()

print(attack_source_counts)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=attack_source_counts.values,
    y=attack_source_counts.index
)

plt.title('Common Sources of Cyberattacks')
plt.xlabel('Number of Attacks')
plt.ylabel('Attack Source')

plt.show()
```

Conclusion

What are the most common types of security vulnerabilities?

```python
# Count vulnerabilities per year
vulnerability_year = (
    df.groupby(['Year', 'Security Vulnerability Type'])
      .size()
      .reset_index(name='Number of Incidents')
)

# Display the data
print(vulnerability_year)

# Create scatter plot
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=vulnerability_year,
    x='Year',
    y='Number of Incidents',
    hue='Security Vulnerability Type',
    s=150
)

plt.title('Security Vulnerabilities by Year')
plt.xlabel('Year')
plt.ylabel('Number of Incidents')

plt.xticks(sorted(df['Year'].unique()))

plt.legend(
    title='Security Vulnerability Type',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

```python
vulnerability_counts = df[
    'Security Vulnerability Type'
].value_counts()

print(vulnerability_counts)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=vulnerability_counts.values,
    y=vulnerability_counts.index
)

plt.title('Common Security Vulnerabilities')
plt.xlabel('Number of Incidents')
plt.ylabel('Security Vulnerability')

plt.show()
```

Conclusion

What security tools are commonly used by industries?

```python
defense_counts = df[
    'Defense Mechanism Used'
].value_counts()

print(defense_counts)

security_tools = pd.crosstab(
    df['Target Industry'],
    df['Defense Mechanism Used']
)

security_tools


plt.figure(figsize=(12, 7))

sns.heatmap(
    security_tools,
    annot=True,
    fmt='d',
    cmap='YlGnBu'
)

plt.title('Security Tools Used Across Industries')
plt.xlabel('Defense Mechanism')
plt.ylabel('Industry')

plt.tight_layout()
plt.show()
```

Conclusion

How many users are affected by cyberattacks?


```python
users_year = (
    df.groupby('Year')['Number of Affected Users']
    .sum()
)

total_users = df['Number of Affected Users'].sum()

print("Total affected-user instances:",
      f"{total_users:,}")

plt.figure(figsize=(11, 6))

plt.fill_between(
    users_year.index,
    users_year.values,
    alpha=0.4
)

plt.plot(
    users_year.index,
    users_year.values,
    marker='o'
)

plt.title(
    f'Affected Users from Cyberattacks by Year\n'
    f'Total: {total_users:,}'
)

plt.xlabel('Year')
plt.ylabel('Number of Affected Users')

plt.xticks(users_year.index)

plt.tight_layout()
plt.show()
```

```python
total_affected_users = df[
    'Number of Affected Users'
].sum()

average_affected_users = df[
    'Number of Affected Users'
].mean()

median_affected_users = df[
    'Number of Affected Users'
].median()

print("Total affected users:", total_affected_users)
print("Average affected users per incident:",
      round(average_affected_users))
print("Median affected users per incident:",
      round(median_affected_users))
```

How does resolution time affect financial loss?

```python
correlation = df[
    'Incident Resolution Time (in Hours)'
].corr(
    df['Financial Loss (in Million $)']
)

print(
    "Correlation:",
    round(correlation, 3)
)
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x='Incident Resolution Time (in Hours)',
    y='Financial Loss (in Million $)',
    scatter_kws={'alpha': 0.4}
)

plt.title(
    'Incident Resolution Time vs Financial Loss'
)

plt.xlabel('Incident Resolution Time (Hours)')
plt.ylabel('Financial Loss (Million $)')

plt.show()
```
